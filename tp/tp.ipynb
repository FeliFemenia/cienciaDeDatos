```python
#Importacion del dataset

import pandas as pd

df = pd.read_csv("ia_tweets.csv")
```

```python
#Analisis exploratorio

tw_neutrales = df[df['polarity'] == 'NEU']
tw_negativos = df[df['polarity'] == 'N']
tw_positivos = df[df['polarity'] == 'P']

print("Cantidad de tweets en el dataset: ", len(df))
print("Columnas disponibles: ", df.columns)
print("Neutrales: ", len(tw_neutrales))
print("Negativos: ", len(tw_negativos))
print("Positivos: ", len(tw_positivos))
print("Media de likes: ", df["favorite_count"].mean())
print("Maximo de likes: ", df["favorite_count"].max())
```

```python
import matplotlib.pyplot as plt

colores = {
    "P": "green",
    "NEU": "gray",
    "N": "red"
}

df = df[df['favorite_count'] != 2243]

for clase, color in colores.items():
    datos = df[df["polarity"] == clase]

    plt.scatter(
        datos.ID,
        datos["favorite_count"],
        color=color,
        label=clase,
        alpha=0.6
    )

plt.xlabel("Tweet")
plt.ylabel("Cantidad de likes")
plt.title("Likes por tweet según sentimiento")
plt.legend()
plt.show()
```
